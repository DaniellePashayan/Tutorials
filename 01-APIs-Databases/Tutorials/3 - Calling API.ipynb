# Calling an API

In order to make an API call, we cna use the python "requests" library to make different API calls.

Every API is set up differently, so reading the API's documentation is extremely important.

For this example, we will be using a public mock API. Documentation can be found here: https://restful-api.dev/

```python
import requests

# APIs will have a 'base url' that is prefixed with every request. As you saw from setting up your API, you only define the endpoints, which are then automatically attached to your base URL. your base URL in the development environment was 127.0.0.1:8000
BASE_URL = 'https://api.restful-api.dev'

# to make a get request, we can do the following format:
response = requests.get(f'{BASE_URL}/endpoint')

# based on the documentation, we can make a call to the "objects" endpoint to get a list of all the mock data
response = requests.get(f'{BASE_URL}/objects')
print(response)
```

```python
# when we print the response, you can see we get a Response object with a status code indicating the result of the request. so how do we get the actually data?
data = response.json()
print(data)
```

```python
# the requests library has built in functions to help handle errors

response = requests.get(f'{BASE_URL}/objects')
response.raise_for_status()

# the raise for status method will raise an HTTPError if the HTTP request returned an unsuccessful status code. this helps in handling errors effectively since it allows you to then use a try except block to catch and handle the error.

try:
    response = requests.get(f'{BASE_URL}/object')
    response.raise_for_status()
    data = response.json()
    print(data)
except requests.exceptions.HTTPError as err:
    print(f'HTTP error occurred: {err}')
except Exception as err:
    print(f'Other error occurred: {err}')
    
# when you run the above, you will get a 401 error, since the endpoint does not exist. this then triggers the HTTPError, which is then caught by the try except block.
```

```python
# some other methods / attributes of the response object
print(f'{response.status_code=}')
print('-------------------------')
print(f'{response.headers=}')
print('-------------------------')
print(f'{response.text=}')
print('-------------------------')
print(f'{response.content=}')
```

```python
# if you notice the headers, youll see that we didnt include any of the information in our request. So how did the API get it? With each call you make, there are default headers that are sent along with your request. These headers tell the server information about the client making the request, such as the user agent (how the request is being sent), accepted response formats, and more.

# if you were to integrate your API with a front end/website, youd also be able to see information about the browser, the operating system of the user, the user's screen size, and if the website requires any authentication, it will also include authentication headers likes cookies or tokens.

# this helps developers troubleshoot compatability issues with the UI. Chromium browsers (Chrome, Edge, etc) render things different than other browsers like Firefox, Opera, etc.

# example header from browser: 
# User-Agent: Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/153.0.0.0 Safari/537.36

# if the API requires you to include specific headers (like an API key), you would need to include them in your request like this:
headers = {
    "Authorization": "Bearer YOUR_API_KEY"
}
response = requests.get("https://api.example.com/data", headers=headers)

# by default, headers are ADDED to the default headers when passed in via parameter by the requests library
```

```python
# since we got a list of all the objects with our first call, lets get details for a specific object

object_id = "1"

# according to the documentation, we can do so by calling the 'objects/{id}' endpoint
response = requests.get(f"{BASE_URL}/objects/{object_id}")
response.json()
```

```python
# lets add something new to the API by making a post request
# according to the documentation, our request needs to be in this format:
# {
#   "name": "Apple MacBook Pro 16",
#   "data": {
#     "year": 2019,
#     "price": 1849.99,
#     "CPU model": "Intel Core i9",
#     "Hard disk size": "1 TB"
#   }
# }

# if you notice, there is no "id" value being passed in. This means that in their database, the ID is likely being generated automatically when a new object is created.

# lets mock some data and store it in the data variable
data = {
    "name": "Lenovo T490",
    "data": {
        "year": 2026,
        "price": 1899.99,
        "CPU model": "Unknown",
        "Hard disk size": "1 TB"
    }
}

# if you notice, the URL is the same as the one we used for the GET request, but this time we are making a POST request to create a new object
# we are passing in the data variable as the JSON payload
response = requests.post(f"{BASE_URL}/objects", json=data)
response.json()
```

```python
# when we print the response, you can see the newly created object with its automatically generated ID
# just for testing, lets now call the GET endpoint and see our newly added item and ensure it was added correctly
generated_id = response.json().get("id")

response = requests.get(f"{BASE_URL}/objects/{generated_id}")
response.json()
```

```python
# lets say we now know the CPU model and want to update it. we would make a PUT request to do so

data = {
    "name": "Lenovo T490",
    "data": {
        "year": 2026,
        "price": 1899.99,
        "CPU model": "Intel Core i9",
        "Hard disk size": "1 TB"
    }
}

response = requests.put(f"{BASE_URL}/objects/{generated_id}", json=data)

# you can see from the response that the CPU model has now been updated
# with this API, whatever you pass in the data will replace the existing data
# with some other APIs, you may only need to pass in what you are changing. this is why the documentation is important, as functionality between APIs will vary
# with this API, they use a "PATCH" request to update only specific fields of an existing object, and a PUT request to replace the entire object

response.json()
```

```python
# lets now delete the newly created object

response = requests.delete(f"{BASE_URL}/objects/{generated_id}")

# there is actually a lot of controversy among developers about whether a DELETE request should return a 200 response or a 204 response (No Content). But any response that begins with a 2 (e.g., 200, 201, 204) indicates success
response.status_code
```
